# 2. Dataset Description & Feature Definitions

## Dataset Overview

The project uses historical Lending Club loan data covering the period from **2007 to 2015**.

The dataset contains information related to borrowers, loan characteristics, credit history, and financial indicators.

The objective is to use these historical features to predict whether a loan will default.

## Dataset Features

### 1. credit.policy
Indicates whether the customer meets the credit underwriting criteria of LendingClub.com.

- `1` = Customer meets the credit underwriting criteria
- `0` = Customer does not meet the credit underwriting criteria

### 2. purpose
The purpose of the loan.

Possible categories include:

- `credit_card`
- `debt_consolidation`
- `educational`
- `major_purchase`
- `small_business`
- `all_other`

### 3. int.rate
The interest rate of the loan, represented as a proportion.

For example, an interest rate of 11% is stored as `0.11`.

Higher interest rates indicate borrowers judged by LendingClub.com to be more risky.

### 4. installment
The monthly installment owed by the borrower if the loan is funded.

### 5. log.annual.inc
The natural logarithm of the borrower's self-reported annual income.

### 6. dti
The borrower's debt-to-income ratio.

It represents the amount of debt relative to annual income.

### 7. fico
The FICO credit score of the borrower.

### 8. days.with.cr.line
The number of days the borrower has had a credit line.

### 9. revol.bal
The borrower's revolving balance.

This represents the amount unpaid at the end of the credit card billing cycle.

### 10. revol.util
The borrower's revolving line utilization rate.

It represents the amount of the available credit line being used.

### 11. inq.last.6mths
The number of inquiries made by creditors during the last six months.

### 12. delinq.2yrs
The number of times the borrower has been 30 or more days past due on a payment during the previous two years.

### 13. pub.rec
The number of derogatory public records associated with the borrower, such as:

- Bankruptcy filings
- Tax liens
- Judgments

The project objective is to predict whether a loan will default.

The exact target-column name and its values will be verified after loading the dataset.

## Outlier Analysis

Outlier analysis is performed to identify extreme observations in the numerical variables. This is particularly important for variables with strongly skewed distributions, as extreme values can affect scaling and model training.

```python
# Calculate the IQR and number of potential outliers
outlier_summary = []

for column in numerical_features:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = ((df[column] < lower_bound) |
                (df[column] > upper_bound)).sum()

    outlier_summary.append({
        'Feature': column,
        'Q1': Q1,
        'Q3': Q3,
        'IQR': IQR,
        'Lower Bound': lower_bound,
        'Upper Bound': upper_bound,
        'Outlier Count': outliers,
        'Outlier %': (outliers / len(df)) * 100
    })

outlier_summary = pd.DataFrame(outlier_summary)

outlier_summary.sort_values(
    by='Outlier %',
    ascending=False
)
```

- `major_purchase` loans have the lowest default rate at approximately **11.23%** among the categories shown.

This indicates that the purpose of a loan is associated with different levels of repayment risk. Therefore, `purpose` should be retained as a predictive feature and appropriately encoded before building the machine-learning model.

```python
# Step 6 — Train/Test Split

from sklearn.model_selection import train_test_split

# Separate features and target
X = df.drop('not.fully.paid', axis=1)
y = df['not.fully.paid']

# Stratified 80/20 split
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training set shape:", X_train.shape)
print("Test set shape:", X_test.shape)

print("\nTraining target distribution:")
print(y_train.value_counts())
print(y_train.value_counts(normalize=True))

print("\nTest target distribution:")
print(y_test.value_counts())
print(y_test.value_counts(normalize=True))
```

### Observation – Step 6: Train/Test Split

- The dataset was divided into **80% training data and 20% testing data** using a stratified train/test split.
- The training set contains **7,662 records with 13 features**, while the test set contains **1,916 records with 13 features**.
- The target distribution is almost identical in both sets.
- In the training set, **83.99%** of loans are fully paid and **16.01%** are not fully paid.
- In the test set, **83.97%** of loans are fully paid and **16.03%** are not fully paid.
- The similar target proportions indicate that **stratification preserved the class distribution** between the training and test sets.
- Therefore, the data is appropriately prepared for building and evaluating machine-learning models.

### Step 7 — Preprocess Features

```python
# Step 7 — Feature Preprocessing

from sklearn.preprocessing import StandardScaler

# Identify categorical and numerical features
categorical_features = ['purpose']

numerical_features = [
    col for col in X_train.columns
    if col not in categorical_features
]

# One-hot encode the categorical feature
X_train_encoded = pd.get_dummies(
    X_train,
    columns=categorical_features,
    drop_first=True
)

X_test_encoded = pd.get_dummies(
    X_test,
    columns=categorical_features,
    drop_first=True
)

# Make sure train and test have the same columns
X_test_encoded = X_test_encoded.reindex(
    columns=X_train_encoded.columns,
    fill_value=0
)

# Scale numerical features
scaler = StandardScaler()

X_train_encoded[numerical_features] = scaler.fit_transform(
    X_train_encoded[numerical_features]
)

X_test_encoded[numerical_features] = scaler.transform(
    X_test_encoded[numerical_features]
)

print("Training features shape:", X_train_encoded.shape)
print("Testing features shape:", X_test_encoded.shape)

print("\nPreprocessing completed successfully.")
```

### Observation – Step 7: Feature Preprocessing

- The categorical feature **purpose** was converted into numerical dummy variables using one-hot encoding.
- The numerical features were standardized using **StandardScaler** so that they are on a comparable scale.
- The preprocessing increased the number of features from **13 to 18**.
- The training set contains **7,662 observations and 18 features**, while the test set contains **1,916 observations and 18 features**.
- The same feature structure was maintained for both training and testing datasets.
- The preprocessing was completed successfully, making the data suitable for machine-learning model training.

### Step 8 — Model Training

Since this is a binary classification problem (not.fully.paid), let’s start with Logistic Regression as our baseline model.

```python
# Step 8 – Train Logistic Regression Model

from sklearn.linear_model import LogisticRegression

# Create the logistic regression model
logistic_model = LogisticRegression(
    max_iter=1000,
    random_state=42
)

# Train the model
logistic_model.fit(X_train_encoded, y_train)

print("Logistic Regression model trained successfully.")
```

## Observation – Step 8: Model Training

- A Logistic Regression model was selected as the baseline classification model.
- The model was trained using the preprocessed training dataset.
- The training was completed successfully without errors.
- The model is now ready to make predictions on the unseen test dataset.

### Step 9 — Model Prediction and Evaluation
Now we will use the trained Logistic Regression model to predict the test set and calculate the basic evaluation metrics.

```python
# Step 9 – Model Prediction and Evaluation

from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

# Make predictions on the test set
y_pred = logistic_model.predict(X_test_encoded)

# Calculate accuracy
accuracy = accuracy_score(y_test, y_pred)

print("Model Accuracy:", round(accuracy, 4))

# Confusion matrix
print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))

# Classification report
print("\nClassification Report:")
print(classification_report(y_test, y_pred))
```

## Observation – Step 9: Model Prediction and Evaluation

- The Logistic Regression model achieved an overall accuracy of **84.03%** on the test dataset.
- For fully paid loans (Class 0), the model performed well, with a recall of **99%** and an F1-score of **0.91**.
- For not fully paid loans (Class 1), the model showed weak performance, with a recall of only **5%** and an F1-score of **0.08**.
- The confusion matrix shows that **293 out of 307 not fully paid loans were incorrectly classified as fully paid**.
- Therefore, although the overall accuracy is relatively high, the model is not effective at identifying loans that may not be fully paid.
- This indicates that additional modelling approaches may be needed to improve the detection of the minority class.

```python
# Step 10 – Train Random Forest Model

from sklearn.ensemble import RandomForestClassifier

# Create the Random Forest model
random_forest_model = RandomForestClassifier(
    n_estimators=200,
    random_state=42,
    class_weight="balanced"
)

# Train the model
random_forest_model.fit(X_train_encoded, y_train)

print("Random Forest model trained successfully.")
```

```python
# Step 11 – Random Forest Prediction and Evaluation

# Make predictions using the Random Forest model
y_pred_rf = random_forest_model.predict(X_test_encoded)

# Calculate accuracy
accuracy_rf = accuracy_score(y_test, y_pred_rf)

print("Random Forest Accuracy:", round(accuracy_rf, 4))

# Confusion matrix
print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred_rf))

# Classification report
print("\nClassification Report:")
print(classification_report(y_test, y_pred_rf))
```

## Observation – Step 11: Random Forest Model Evaluation

- The Random Forest model achieved an overall accuracy of **83.98%** on the test dataset.
- For fully paid loans (Class 0), the model performed very well, with a recall of **100%** and an F1-score of **0.91**.
- For not fully paid loans (Class 1), the model performed poorly, with a recall of only **1%** and an F1-score of **0.02**.
- The confusion matrix shows that **304 out of 307 not fully paid loans were incorrectly classified as fully paid**.
- Although the overall accuracy is relatively high, the model is not effective at identifying loans that may not be fully paid.
- The Random Forest model therefore does not improve the detection of the minority class compared with Logistic Regression.

```python
# Step 12 – Model Comparison

import pandas as pd

# Create a comparison table
model_comparison = pd.DataFrame({
    "Model": ["Logistic Regression", "Random Forest"],
    "Accuracy": [accuracy, accuracy_rf],
    "Class 1 Precision": [0.52, 0.50],
    "Class 1 Recall": [0.05, 0.01],
    "Class 1 F1-Score": [0.08, 0.02]
})

print("Model Performance Comparison:")
display(model_comparison)
```

## Observation – Step 12: Model Comparison

- Logistic Regression achieved an accuracy of **84.03%**, while Random Forest achieved an accuracy of **83.98%**.
- The overall accuracy of both models is therefore very similar.
- For Class 1 (Not Fully Paid), Logistic Regression performed better, with a precision of **0.52**, recall of **0.05**, and F1-score of **0.08**.
- Random Forest performed worse for Class 1, with a precision of **0.50**, recall of only **0.01**, and F1-score of **0.02**.
- Therefore, **Logistic Regression is the better-performing model of the two for identifying not fully paid loans**.
- However, the low recall for Class 1 indicates that both models still have difficulty detecting loans that may not be fully paid.

### Step 13 — Balanced Logistic Regression
The previous Logistic Regression model had good overall accuracy but detected very few Class 1 (Not Fully Paid) loans. Here we'll use class_weight="balanced" so the minority class receives more importance during training.

```python
# Step 13 – Balanced Logistic Regression

from sklearn.linear_model import LogisticRegression

# Create a balanced Logistic Regression model
balanced_logistic_model = LogisticRegression(
    max_iter=1000,
    random_state=42,
    class_weight="balanced"
)

# Train the model
balanced_logistic_model.fit(X_train_encoded, y_train)

print("Balanced Logistic Regression model trained successfully.")
```

```python
# Step 14 — Evaluate Balanced Logistic Regression

# Make predictions on the test set
y_pred_balanced = balanced_logistic_model.predict(X_test_encoded)

# Calculate accuracy
accuracy_balanced = accuracy_score(y_test, y_pred_balanced)

print("Balanced Logistic Regression Accuracy:",
      round(accuracy_balanced, 4))

# Confusion matrix
print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred_balanced))

# Classification report
print("\nClassification Report:")
print(classification_report(y_test, y_pred_balanced))
```

```python
# Step 15 — Final Model Comparison

final_model_comparison = pd.DataFrame({
    "Model": [
        "Logistic Regression",
        "Random Forest",
        "Balanced Logistic Regression"
    ],
    "Accuracy": [
        accuracy,
        accuracy_rf,
        accuracy_balanced
    ],
    "Class 1 Precision": [
        0.52,
        0.50,
        0.24
    ],
    "Class 1 Recall": [
        0.05,
        0.01,
        0.57
    ],
    "Class 1 F1-Score": [
        0.08,
        0.02,
        0.34
    ]
})

print("Final Model Performance Comparison:")
display(final_model_comparison)
```

## Observation – Step 16: Final Model Selection

• Logistic Regression achieved the highest overall accuracy at 84.03%, but its recall for Class 1 (Not Fully Paid) was only 5%.

• Random Forest achieved a very similar accuracy of 83.98%, but its Class 1 recall was only 1%, making it the weakest model for identifying potentially not fully paid loans.

• Balanced Logistic Regression achieved a lower overall accuracy of 64.04%, but substantially improved Class 1 recall to 57% and the Class 1 F1-score to 0.34.

• The balanced model correctly identified 175 out of 307 Class 1 loans, compared with only 14 identified by the original Logistic Regression model.

• Therefore, if the main objective is to identify loans that may not be fully paid, Balanced Logistic Regression is the preferred model.

• The results demonstrate the trade-off between overall accuracy and minority-class detection in an imbalanced classification problem.

```python
# Step 17 — ROC Curve and AUC

from sklearn.metrics import roc_curve, roc_auc_score
import matplotlib.pyplot as plt

# Get prediction probabilities for Class 1
y_prob_logistic = logistic_model.predict_proba(X_test_encoded)[:, 1]
y_prob_rf = random_forest_model.predict_proba(X_test_encoded)[:, 1]
y_prob_balanced = balanced_logistic_model.predict_proba(X_test_encoded)[:, 1]

# Calculate ROC curves
fpr_logistic, tpr_logistic, _ = roc_curve(y_test, y_prob_logistic)
fpr_rf, tpr_rf, _ = roc_curve(y_test, y_prob_rf)
fpr_balanced, tpr_balanced, _ = roc_curve(y_test, y_prob_balanced)

# Calculate AUC scores
auc_logistic = roc_auc_score(y_test, y_prob_logistic)
auc_rf = roc_auc_score(y_test, y_prob_rf)
auc_balanced = roc_auc_score(y_test, y_prob_balanced)

print("ROC-AUC Scores:")
print("Logistic Regression:", round(auc_logistic, 4))
print("Random Forest:", round(auc_rf, 4))
print("Balanced Logistic Regression:", round(auc_balanced, 4))

# Plot ROC curves
plt.figure(figsize=(8, 6))

plt.plot(
    fpr_logistic,
    tpr_logistic,
    label=f"Logistic Regression (AUC = {auc_logistic:.3f})"
)

plt.plot(
    fpr_rf,
    tpr_rf,
    label=f"Random Forest (AUC = {auc_rf:.3f})"
)

plt.plot(
    fpr_balanced,
    tpr_balanced,
    label=f"Balanced Logistic Regression (AUC = {auc_balanced:.3f})"
)

plt.plot([0, 1], [0, 1], linestyle="--", label="Random Guess")

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve Comparison")
plt.legend()
plt.grid(True)
plt.show()
```

### Observation – ROC Curve Comparison
- The ROC curves show that all three models perform better than random guessing, with AUC values above 0.50.
- Logistic Regression achieved an AUC of 0.6835, indicating moderate ability to distinguish between fully paid and not fully paid loans.
- Random Forest achieved a lower AUC of 0.6688, making it the weakest of the three models based on ROC-AUC.
- Balanced Logistic Regression achieved the highest AUC of 0.6845, slightly outperforming the original Logistic Regression.
- The ROC-AUC results are very similar for the two Logistic Regression models, suggesting that balancing the classes improves minority-class detection without substantially changing overall discrimination ability.
- However, the Balanced Logistic Regression remains the preferred model because its Class 1 recall increased substantially to 57%, which is more important for identifying loans that may not be fully paid.

# Step 18 – Final Conclusion and Recommendations

## Conclusion

This analysis evaluated Lending Club loan data to identify loans that may not be fully paid. Several classification approaches were compared, including Logistic Regression, Random Forest, and Balanced Logistic Regression.

The original Logistic Regression model achieved the highest overall accuracy of approximately 84.03%, while Random Forest achieved approximately 83.98%. However, both models performed poorly in identifying the minority class of loans that were not fully paid.

Balanced Logistic Regression produced a lower overall accuracy of approximately 64.04%, but it substantially improved the detection of Class 1 loans. Its Class 1 recall increased to 57%, compared with only 5% for the original Logistic Regression and 1% for Random Forest. Its Class 1 F1-score also improved to 0.34.

The ROC-AUC results were similar across the models. Balanced Logistic Regression achieved the highest AUC of 0.6845, followed closely by Logistic Regression at 0.6835, while Random Forest achieved 0.6688.

## Final Model Selection

Based on the objective of identifying loans that may not be fully paid, Balanced Logistic Regression is selected as the preferred model. Although it sacrifices overall accuracy, it is substantially better at identifying the minority class.

## Recommendations

- Use Balanced Logistic Regression when identifying potentially risky loans is the primary objective.
- Give greater importance to minority-class recall rather than relying only on overall accuracy.
- Consider additional techniques such as threshold tuning, resampling, or alternative classification algorithms to further improve minority-class detection.
- Future analysis could include additional feature engineering and hyperparameter tuning to improve predictive performance.

## Overall Finding

The analysis demonstrates the trade-off between overall accuracy and minority-class detection in an imbalanced loan classification problem. A model with lower overall accuracy can be more useful when it substantially improves the identification of potentially risky loans.
